```python
import numpy as np
import pandas as pd


df = pd.read_csv('/content/delhi-weather-aqi-2025 (1).csv')
null_counts = df.isnull().sum()
print(null_counts)
print("Total missing:", null_counts.sum())
```

```python
np.random.seed(42)
df_missing = df.copy()
rows_t = np.random.choice(df_missing.index, 150, replace=False)
rows_a = np.random.choice(df_missing.index, 100, replace=False)
df_missing.loc[rows_t, 'temp_c'] = np.nan
df_missing.loc[rows_a, 'aqi_index'] = np.nan


temp_mean = df_missing['temp_c'].mean()
aqi_mean = df_missing['aqi_index'].mean()
df_missing.fillna({'temp_c': temp_mean, 'aqi_index': aqi_mean}, inplace=True)


print(df_missing.isnull().sum()[['temp_c', 'aqi_index']])
```

```python
top8_hottest = df.sort_values(by='temp_c', ascending=False).head(8)
print(top8_hottest[['date_ist', 'time_ist', 'location', 'temp_c']])


anand_vihar = df[df['location'] == 'Anand Vihar']
top5_polluted_av = anand_vihar.sort_values(by='aqi_index', ascending=False).head(5)
print(top5_polluted_av[['date_ist', 'time_ist', 'aqi_index', 'pm2_5']])
```

```python
print(df['location'].value_counts())
print(df['condition_text'].value_counts())
```

```python
print(df.sort_index(ascending=False).head(3)[['location', 'temp_c']])

# Recreate the datetime column as df was overwritten by a fresh read_csv in a previous cell
if 'datetime' not in df.columns:
    df['datetime'] = pd.to_datetime(df['date_ist'] + ' ' + df['time_ist'], format='%d/%m/%Y %H:%M')

df_explicit = df.set_index('datetime')
print(df_explicit.sort_index().head(3)[['location', 'temp_c']])
```

```python
case1 = df.loc[(df['temp_c'] > 40) & (df['location'] == 'Anand Vihar'),
               ['date_ist', 'time_ist', 'temp_c']]


case2 = df.loc[(df['aqi_index'] > 300) & (df['pm2_5'] > 200),
               ['location', 'aqi_index', 'pm2_5']]


case3 = df.loc[(df['humidity'] < 20) & (df['windspeed_kph'] > 10),
               ['date_ist', 'location', 'humidity', 'windspeed_kph']]


print(case1.head()); print(case2.head()); print(case3.head())
```

```python
print("Max:", df['temp_c'].max(), "Min:", df['temp_c'].min())
print(df.loc[df['temp_c'].idxmax()])
print("Max AQI:", df['aqi_index'].max(), "Min AQI:", df['aqi_index'].min())
```

```python
avg_temp_by_location = df.groupby('location')['temp_c'].mean().sort_values(ascending=False)
print(avg_temp_by_location)


df['month'] = df['datetime'].dt.month_name()
avg_aqi_loc_month = df.groupby(['location', 'month'])['aqi_index'].mean()
print(avg_aqi_loc_month.head(6))
```

```python
df['heatwave_alert'] = np.where(df['temp_c'] >= 40, 'Yes', 'No')
print(df['heatwave_alert'].value_counts())
```

```python
agg1 = df.groupby('location').agg({'temp_c': 'mean', 'aqi_index': 'max'})
print(agg1)


agg2 = df.groupby('month').agg({'temp_c': ['mean', 'max', 'min']})
print(agg2.head(4))
```

```python
grouped = df.groupby('location')
anand_vihar_group = grouped.get_group('Anand Vihar')
print(anand_vihar_group.shape)

```

```python
print(df['temp_c'].corr(df['aqi_index']))
print(df[['temp_c', 'humidity', 'aqi_index', 'pm2_5', 'pm10']].corr())
```

```python
df['temp_c_normalized'] = (df['temp_c'] - df['temp_c'].min()) / (df['temp_c'].max() - df['temp_c'].min())
print(df[['temp_c', 'temp_c_normalized']].head())


df['aqi_zscore'] = (df['aqi_index'] - df['aqi_index'].mean()) / df['aqi_index'].std()
```

```python
weather_part = df[['date_ist', 'time_ist', 'location', 'temp_c', 'humidity']]
pollution_part = df[['date_ist', 'time_ist', 'location', 'aqi_index', 'pm2_5', 'pm10']]


merged_df = pd.merge(weather_part, pollution_part, on=['date_ist', 'time_ist', 'location'], how='inner')
print(merged_df.shape)


concatenated_df = pd.concat([df.iloc[:100], df.iloc[100:200]], axis=0)
print(concatenated_df.shape)
```
